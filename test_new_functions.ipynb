```python
#### STEP 01 ####
## [func] Resample T1w and MT- to 0.5mm, antsResampleImage
# Input: Native T1w/MT-
# Output: T1w/MT- on 0.5mm isotrpic grid

from pathlib import Path

import ants
import nibabel as nib
import matplotlib.pyplot as plt

def get_image_geometry(path):
    img = nib.load(path)

    shape = img.shape[:3]
    voxel_size = img.header.get_zooms()[:3]
    physical_size = tuple(round(s * v, 3) for s, v in zip(shape, voxel_size))

    return {
        "shape": shape,
        "voxel_size": voxel_size,
        "physical_size_mm": physical_size,
        "orientation": nib.aff2axcodes(img.affine),
    }

def resample_image(
    input_path:str,
    output_path: str,
    #mode="isotropic05", #for future changes
    target_spacing=(0.5, 0.5, 0.5),
    interp_type: int=1, #0 (linear), 1 (nearest neighbor), 2 (gaussian), 3 (windowed sinc), 4 (bspline)
):

    input_path = Path(input_path)
    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    img = ants.image_read(str(input_path))

    resampled = ants.resample_image(
        img,
        resample_params=target_spacing,
        use_voxels=False,
        interp_type=int(interp_type),
    )
    ants.image_write(resampled, str(output_path))

    print("=== Resampling Information ===")
    print(f"Input : {input_path}")
    print(f"Output: {output_path}")
    #print(f"Mode  : {mode}")
    print(f"Interpolation: {interp_type}")

    print("\nOriginal Image:")
    print(get_image_geometry(input_path))

    print("\nResampled Image:")
    print(get_image_geometry(output_path))

    return output_path
```

```python
from __future__ import annotations

import os
import shutil
import subprocess
from pathlib import Path


def build_group_template(
    image_paths,
    output_dir,
    template_name="T1w", #MTminus
    iterations=6,
    threads=8,
    overwrite=False,
):

    image_paths = [Path(path).resolve() for path in image_paths]
    output_dir = Path(output_dir).resolve()

    if len(image_paths) < 2:
        raise ValueError(
            "At least two images are needed to build a group template."
        )

    for path in image_paths:
        if not path.exists():
            raise FileNotFoundError(f"Image not found: {path}")

    ants_script = shutil.which(
        "antsMultivariateTemplateConstruction2.sh"
    )

    if ants_script is None:
        raise RuntimeError(
            "antsMultivariateTemplateConstruction2.sh was not found.\n"
            "Check it with:\n"
            "which antsMultivariateTemplateConstruction2.sh"
        )

    work_dir = output_dir / f"{template_name}_group_template"
    input_dir = work_dir / "inputs"

    output_prefix = work_dir / f"{template_name}_"

    template_path = (
        work_dir / f"{template_name}_template0.nii.gz"
    )

    log_path = work_dir / "template_construction.log"

    if template_path.exists() and not overwrite:
        print(f"Template already exists: {template_path}")

        return {
            "template": str(template_path),
            "log": str(log_path),
        }

    if work_dir.exists() and overwrite:
        shutil.rmtree(work_dir)

    input_dir.mkdir(parents=True, exist_ok=True)

    staged_paths = []

    for index, source_path in enumerate(image_paths):
        staged_path = (
            input_dir
            / f"subject_{index:04d}_{template_name}.nii.gz"
        )

        staged_path.symlink_to(source_path)
        staged_paths.append(staged_path)

    command = [
        ants_script,
        "-d", "3",
        "-i", str(iterations),
        "-k", "1",
        "-n", "1",
        "-r", "1",
        "-l", "1",
        "-t", "SyN", #Greedy SyN
        "-m", "CC", # cross-correlation
        "-c", "0",
        "-o", str(output_prefix),
        *[str(path) for path in staged_paths],
    ]

    environment = os.environ.copy()
    environment["ITK_GLOBAL_DEFAULT_NUMBER_OF_THREADS"] = str(
        threads
    )

    print("\n=== Group-template construction ===")
    print(f"Modality        : {template_name}")
    print(f"Number of images: {len(image_paths)}")
    print(f"Iterations      : {iterations}")
    print(f"Threads         : {threads}")
    print(f"Expected output : {template_path}")
    print()

    with log_path.open("w") as log_file:
        process = subprocess.Popen(
            command,
            cwd=str(input_dir),
            env=environment,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
        )

        if process.stdout is None:
            raise RuntimeError("Could not read ANTs output.")

        for line in process.stdout:
            print(line, end="")
            log_file.write(line)
            log_file.flush()

        return_code = process.wait()

    if return_code != 0:
        raise RuntimeError(
            "Template construction failed.\n"
            f"Check the log: {log_path}"
        )

    if not template_path.exists():
        raise RuntimeError(
            "ANTs finished, but the expected template was not found:\n"
            f"{template_path}"
        )

    print("\nTemplate construction finished.")
    print(f"Template: {template_path}")
    print(f"Log     : {log_path}")

    return {
        "template": str(template_path),
        "log": str(log_path),
        "input_images": [str(path) for path in image_paths],
        "template_name": template_name,
        "iterations": iterations,
    }
```

```python
#### STEP 02 ####
## [func] TSE/MT+ register/align to template-space T1w/MT-, antsRegistrationSyNQuick
# Input: fixed_img=T1w/MT- templates and moving_img=TSE/MT+
# Output: Aligned TSE/MT+ and transformation

import json
from pathlib import Path

import ants


def register_syn(
    fixed_path,
    moving_path,
    output_dir,
    *,
    output_name="moving_in_fixed",
    transform_prefix="transform_",
    type_of_transform="antsRegistrationSyNQuick[s]", #"antsRegistrationSyN[s]"
    overwrite=False,
):

    fixed_path = Path(fixed_path)
    moving_path = Path(moving_path)
    output_dir = Path(output_dir)

    if not fixed_path.exists():
        raise FileNotFoundError(f"Fixed image not found: {fixed_path}")

    if not moving_path.exists():
        raise FileNotFoundError(f"Moving image not found: {moving_path}")

    output_dir.mkdir(parents=True, exist_ok=True)

    warped_path = output_dir / f"{output_name}.nii.gz"
    json_path = output_dir / f"{output_name}_transforms.json"
    ants_prefix = output_dir / transform_prefix

    if warped_path.exists() and json_path.exists() and not overwrite:
        print(f"Already registered: {warped_path}")

        with json_path.open("r") as file:
            return json.load(file)

    print("\n=== Registration ===")
    print(f"Fixed          : {fixed_path}")
    print(f"Moving         : {moving_path}")
    print(f"Transform type : {type_of_transform}")

    fixed = ants.image_read(str(fixed_path)).clone("float")
    moving = ants.image_read(str(moving_path)).clone("float")

    registration = ants.registration(
        fixed=fixed,
        moving=moving,
        type_of_transform=type_of_transform,
        outprefix=str(ants_prefix),
        verbose=True,
    )

    ants.image_write(
        registration["warpedmovout"],
        str(warped_path),
    )

    forward_transforms = [
        str(path) for path in registration["fwdtransforms"]
    ]

    inverse_transforms = [
        str(path) for path in registration["invtransforms"]
    ]

    affine = next(
        (
            path for path in forward_transforms
            if path.endswith("GenericAffine.mat")
        ),
        None,
    )

    forward_warp = next(
        (
            path for path in forward_transforms
            if path.endswith("Warp.nii.gz")
            and not path.endswith("InverseWarp.nii.gz")
        ),
        None,
    )

    inverse_warp = next(
        (
            path for path in inverse_transforms
            if path.endswith("InverseWarp.nii.gz")
        ),
        None,
    )

    info = {
        "fixed": str(fixed_path),
        "moving": str(moving_path),
        "type_of_transform": type_of_transform,
        "warped_image": str(warped_path),
        "forward_transforms": forward_transforms,
        "inverse_transforms": inverse_transforms,
        "affine": affine,
        "forward_warp": forward_warp,
        "inverse_warp": inverse_warp,
    }

    with json_path.open("w") as file:
        json.dump(info, file, indent=2)

    print(f"Saved warped image: {warped_path}")
    print(f"Saved transforms  : {json_path}")

    return info
```

```python
from __future__ import annotations

import os

# Prevent every process from starting several internal numerical threads.
# Set these before importing NumPy.
os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")

from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path

import nibabel as nib
import numpy as np
import pandas as pd


ROOT = Path("/media/maria/A91A-44D3/mri_hiwi/exp_all_subs_ses001")


def compute_contrast(lc_values: np.ndarray, dpt_values: np.ndarray) -> dict:
    """
    Compute LC contrast using mean LC and mean DPT intensity:

        LC contrast = (LC_mean - DPT_mean) / DPT_mean
    """
    lc_values = lc_values[np.isfinite(lc_values)]
    dpt_values = dpt_values[np.isfinite(dpt_values)]

    if lc_values.size == 0:
        return {
            "lc_voxel_count": 0,
            "lc_mean": np.nan,
            "dpt_mean": np.nan,
            "lc_contrast": np.nan,
        }

    if dpt_values.size == 0:
        return {
            "lc_voxel_count": int(lc_values.size),
            "lc_mean": float(np.mean(lc_values)),
            "dpt_mean": np.nan,
            "lc_contrast": np.nan,
        }

    lc_mean = float(np.mean(lc_values))
    dpt_mean = float(np.mean(dpt_values))

    contrast = (
        (lc_mean - dpt_mean) / dpt_mean
        if dpt_mean != 0
        else np.nan
    )

    return {
        "lc_voxel_count": int(lc_values.size),
        "lc_mean": lc_mean,
        "dpt_mean": dpt_mean,
        "lc_contrast": float(contrast),
    }


def process_subject(subject: str, session: str = "ses001") -> list[dict]:
    """
    Process one subject.

    The subject is processed independently so this function can safely run
    inside a ProcessPoolExecutor.
    """
    tse_path = (
        ROOT
        / "step6_tse_mni"
        / subject
        / session
        / "tse_in_MNI_brainstem_0p5mm.nii.gz"
    )

    original_lc_path = (
        ROOT
        / "step7_masks_grid"
        / subject
        / session
        / "LC_mask_grid.nii.gz"
    )

    dpt_path = (
        ROOT
        / "step7_masks_grid"
        / subject
        / session
        / "DPT_mask_grid.nii.gz"
    )

    cluster_lc_path = (
        ROOT
        / "step8_contrast"
        / subject
        / session
        / "lc_ft_cluster_mask.nii.gz"
    )

    required_paths = {
        "TSE": tse_path,
        "Original LC mask": original_lc_path,
        "DPT mask": dpt_path,
        "Cluster LC mask": cluster_lc_path,
    }

    missing = [
        f"{name}: {path}"
        for name, path in required_paths.items()
        if not path.exists()
    ]

    if missing:
        return [{
            "subject": subject,
            "session": session,
            "mask": "error",
            "error": "Missing files: " + "; ".join(missing),
        }]

    # Load the TSE only once.
    # float32 uses half the memory of float64 and is sufficient here.
    tse = np.asarray(
        nib.load(str(tse_path)).dataobj,
        dtype=np.float32,
    )

    # Load the DPT mask only once.
    dpt_mask = np.asarray(
        nib.load(str(dpt_path)).dataobj
    ) > 0

    dpt_values = tse[dpt_mask]

    results = []

    lc_paths = {
        "Original LC mask": original_lc_path,
        "Funnel-tip cluster mask": cluster_lc_path,
    }

    for mask_name, mask_path in lc_paths.items():
        lc_mask = np.asarray(
            nib.load(str(mask_path)).dataobj
        ) > 0

        metrics = compute_contrast(
            lc_values=tse[lc_mask],
            dpt_values=dpt_values,
        )

        results.append({
            "subject": subject,
            "session": session,
            "mask": mask_name,
            **metrics,
            "error": None,
        })

    return results


def find_subjects(session: str = "ses001") -> list[str]:
    """
    Find subjects that have a session folder in step6_tse_mni.
    """
    step6_root = ROOT / "step6_tse_mni"

    return sorted(
        subject_dir.name
        for subject_dir in step6_root.glob("sub*")
        if subject_dir.is_dir()
        and (subject_dir / session).is_dir()
    )


def run_all_subjects_fast(
    session: str = "ses001",
    max_workers: int | None = None,
) -> pd.DataFrame:
    subjects = find_subjects(session=session)

    if not subjects:
        raise RuntimeError(
            f"No subjects found under {ROOT / 'step6_tse_mni'}"
        )

    if max_workers is None:
        # Conservative default for an external drive.
        max_workers = min(4, os.cpu_count() or 1)

    all_results = []

    with ProcessPoolExecutor(max_workers=max_workers) as executor:
        future_to_subject = {
            executor.submit(process_subject, subject, session): subject
            for subject in subjects
        }

        for completed, future in enumerate(
            as_completed(future_to_subject),
            start=1,
        ):
            subject = future_to_subject[future]

            try:
                subject_results = future.result()
                all_results.extend(subject_results)
                print(
                    f"[{completed:03d}/{len(subjects):03d}] "
                    f"Finished {subject}"
                )

            except Exception as error:
                all_results.append({
                    "subject": subject,
                    "session": session,
                    "mask": "error",
                    "error": repr(error),
                })

                print(
                    f"[{completed:03d}/{len(subjects):03d}] "
                    f"ERROR {subject}: {error}"
                )

    df = pd.DataFrame(all_results)

    output_path = ROOT / f"lc_contrast_all_subjects_{session}.csv"
    df.to_csv(output_path, index=False)

    print(f"\nSaved: {output_path}")

    return df


if __name__ == "__main__":
    results_df = run_all_subjects_fast(
        session="ses001",
        max_workers=4,
    )

    print(results_df)
```
